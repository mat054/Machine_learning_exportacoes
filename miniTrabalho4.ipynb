```python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
import numpy as np

arquivo_filtrado = arquivo[['Area', 'Year','Item', 'Import','Export ','Production','Consumption']].reset_index()

ordenados = arquivo_filtrado['Area'].value_counts().sort_values(ascending=False).reset_index()
lista_paises = ordenados['Area'].tolist()[:1]

especiarias = arquivo_filtrado['Item'].unique().tolist()

# Criar uma cópia do dataframe original para construir o resultado final
resultado_final = arquivo_filtrado.copy()

print("VERSÃO 6")

for coluna in arquivo_filtrado.columns:
    # Verificar se é uma coluna numérica e excluir 'Year' e 'index' da análise
    if (arquivo_filtrado[coluna].dtype == 'int64' or arquivo_filtrado[coluna].dtype == 'float64') and coluna != 'Year' and coluna != 'index':
        for pais in lista_paises:
            for especiaria in especiarias:
                print(f"Processando país {pais}, especiaria {especiaria} sobre a coluna {coluna}")
                arquivo_pais = arquivo_filtrado[(arquivo_filtrado['Area'] == pais) & (arquivo_filtrado['Item'] == especiaria)].copy()
                
                # Se o dataframe não estiver vazio e tiver dados suficientes
                if not arquivo_pais.empty and len(arquivo_pais) > 1:
                    # Criar a coluna de outliers usando o método de quartis com janela
                    outlier_flags = []
                    
                    for i in range(len(arquivo_pais)):
                        ano_central = arquivo_pais.iloc[i]['Year']
                        
                        # Janela de 3 anos para cada lado (ou todos os dados disponíveis)
                        janela = arquivo_filtrado[(arquivo_filtrado['Area'] == pais) & 
                                                 (arquivo_filtrado['Item'] == especiaria) &
                                                 (arquivo_filtrado['Year'] >= ano_central - 3) & 
                                                 (arquivo_filtrado['Year'] <= ano_central + 3)]
                        
                        # Calcular quartis apenas se tivermos dados suficientes
                        if len(janela) >= 3:
                            q1 = janela[coluna].quantile(0.25)
                            q3 = janela[coluna].quantile(0.75)
                            iqr = q3 - q1
                            
                            lim_inf = q1 - 1.7 * iqr
                            lim_sup = q3 + 1.7 * iqr
                            
                            valor = arquivo_pais.iloc[i][coluna]
                            outlier_flags.append(valor < lim_inf or valor > lim_sup)
                        else:
                            # Se não há dados suficientes, considerar como não outlier
                            outlier_flags.append(False)
                    
                    # Adicionar a coluna de outliers ao dataframe
                    coluna_outlier = f"{coluna}_is_outlier"
                    arquivo_pais[coluna_outlier] = outlier_flags
                    
                    # Fazer join com o resultado final usando a coluna 'index'
                    # Selecionamos apenas o index e a coluna de outlier para o join
                    df_outlier = arquivo_pais[['index', coluna_outlier]]
                    
                    # CORREÇÃO: Primeiro adicionar a coluna como False no resultado_final 
                    # se essa coluna ainda não existir
                    if coluna_outlier not in resultado_final.columns:
                        resultado_final[coluna_outlier] = False
                    
                    # Fazer o merge utilizando um dicionário para mapear os valores
                    outlier_dict = dict(zip(df_outlier['index'], df_outlier[coluna_outlier]))
                    
                    # Atualizar somente as linhas correspondentes
                    for idx in outlier_dict:
                        resultado_final.loc[resultado_final['index'] == idx, coluna_outlier] = outlier_dict[idx]
                    
                    #Visualização
                    fig, ax = plt.subplots(figsize=(10, 6))
                    
                    #Plotar pontos normais e outliers com cores diferentes
                    normal = arquivo_pais[~arquivo_pais[coluna_outlier]]
                    outliers = arquivo_pais[arquivo_pais[coluna_outlier]]
                    
                    ax.scatter(normal['Year'], normal[coluna], color='blue', label='Normal')
                    ax.scatter(outliers['Year'], outliers[coluna], color='red', label='Outlier')
                    
                    plt.title(f"{pais} - {especiaria} - {coluna}")
                    plt.xlabel('Ano')
                    plt.ylabel(coluna)
                    plt.legend()
                    plt.show()

# Exibir informações sobre o resultado final
print(f"Dataframe final criado com {len(resultado_final)} linhas.")
print(f"Colunas disponíveis: {', '.join(resultado_final.columns)}")

# Opcional: Salvar o resultado
resultado_final.to_csv('resultado_com_outliers.csv', index=False)
```

```python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
import numpy as np

arquivo_filtrado = arquivo[['Area', 'Year','Item', 'Import','Export ','Production','Consumption']]

ordenados = arquivo_filtrado['Area'].value_counts().sort_values(ascending=False).reset_index()
lista_paises = ordenados['Area'].tolist()[:10]

especiarias = arquivo_filtrado['Item'].unique().tolist()

# Lista para armazenar todos os dataframes processados
dataframes_processados = []

for coluna in arquivo_filtrado.columns:
    if (arquivo_filtrado[coluna].dtype == 'int64' or arquivo_filtrado[coluna].dtype == 'float64')and coluna != 'Year':
        for pais in lista_paises:
            for especiaria in especiarias:
                print(f"Processando país {pais}, especiaria {especiaria} sobre a coluna {coluna}")
                arquivo_pais = arquivo_filtrado[(arquivo_filtrado['Area'] == pais) & (arquivo_filtrado['Item'] == especiaria)].copy()
                
                # Se o dataframe não estiver vazio e tiver dados suficientes
                if not arquivo_pais.empty and len(arquivo_pais) > 1:
                    # Criar a coluna de outliers usando o método de quartis com janela
                    outlier_flags = []
                    
                    for i in range(len(arquivo_pais)):
                        ano_central = arquivo_pais.iloc[i]['Year']
                        
                        # Janela de 3 anos para cada lado (ou todos os dados disponíveis)
                        janela = arquivo_filtrado[(arquivo_filtrado['Area'] == pais) & 
                                                 (arquivo_filtrado['Item'] == especiaria) &
                                                 (arquivo_filtrado['Year'] >= ano_central - 3) & 
                                                 (arquivo_filtrado['Year'] <= ano_central + 3)]
                        
                        # Calcular quartis apenas se tivermos dados suficientes
                        if len(janela) >= 3:
                            q1 = janela[coluna].quantile(0.25)
                            q3 = janela[coluna].quantile(0.75)
                            iqr = q3 - q1
                            
                            lim_inf = q1 - 1.5 * iqr
                            lim_sup = q3 + 1.5 * iqr
                            
                            valor = arquivo_pais.iloc[i][coluna]
                            outlier_flags.append(valor < lim_inf or valor > lim_sup)
                        else:
                            # Se não há dados suficientes, considerar como não outlier
                            outlier_flags.append(False)
                    
                    # Adicionar a coluna de outliers ao dataframe
                    coluna_outlier = f"{coluna}_is_outlier"
                    arquivo_pais[coluna_outlier] = outlier_flags
                    
                    # Adicionar ao conjunto de dataframes processados
                    dataframes_processados.append(arquivo_pais)
                    
                    # Visualização
                    fig, ax = plt.subplots(figsize=(10, 6))
                    
                    # Plotar pontos normais e outliers com cores diferentes
                    normal = arquivo_pais[~arquivo_pais[coluna_outlier]]
                    outliers = arquivo_pais[arquivo_pais[coluna_outlier]]
                    
                    ax.scatter(normal['Year'], normal[coluna], color='blue', label='Normal')
                    ax.scatter(outliers['Year'], outliers[coluna], color='red', label='Outlier')
                    
                    plt.title(f"{pais} - {especiaria} - {coluna}")
                    plt.xlabel('Ano')
                    plt.ylabel(coluna)
                    plt.legend()
                    plt.show()

# Concatenar todos os dataframes processados
if dataframes_processados:
    resultado_final = pd.concat(dataframes_processados, ignore_index=True)
    print(f"Dataframe final criado com {len(resultado_final)} linhas.")
    
    # Opcional: Salvar o resultado
    resultado_final.to_csv('resultado_com_outliers.csv', index=False)
else:
    print("Nenhum dataframe foi processado.")
```
